```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
collision_file = np.load(".data/Time_series_colisao.npz")
display(collision_file['TS'].shape)
spreading_file = np.load(".data/Time_series_espalhamento.npz")
display(spreading_file['TS'].shape)
```

```python
height = collision_file['TS'][:, :, 1]
display(height.shape)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 4.5))

ax.set_title("Altura média de todas as simulações")
ax.set_xlabel("Snapshot")
ax.set_ylabel("Altura média")
mean_height = np.mean(height, axis=0)
ax.plot(mean_height)
fig.savefig(".results/mean_height.png", dpi=600)
```

```python
diameter = collision_file['TS'][:, :, 0]
fig, ax = plt.subplots(1, 1, figsize=(8, 4.5))

ax.set_title("Diâmetro médio de todas as simulações")
ax.set_xlabel("Snapshot")
ax.set_ylabel("Diâmetro médio")
mean_diameter = np.mean(diameter, axis=0)
ax.plot(mean_diameter)
fig.savefig(".results/mean_diameter.png", dpi=600)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 4.5), subplot_kw={'projection': '3d'})
x, y = np.meshgrid(np.arange(0, 132), np.arange(0, 1000))
z = spreading_file['TS'][:, :, 1].T
ax.plot_surface(x, y, z, cmap='viridis')
```
