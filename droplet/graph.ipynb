```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
collision_file = np.load(".data/Time_series_colisao.npz")
display(collision_file['TS'].shape)
spreading_file = np.load(".data/Time_series_espalhamento.npz")
display(spreading_file['TS'].shape)
```

```python
cn = ["Diâmetro", "Altura", "Energia Cinética", "Energia Superficial", "Energia Dissipativa"]
cn2 = ["diametro", "altura", "energia_cinetica", "energia_superficial", "energia_dissipativa"]

for i in range(len(cn)):
    fig, ax = plt.subplots(1, 1, figsize=(5, 4))
    mean_component = np.mean(collision_file['TS'][:, :, i], axis=0)
    std_component = np.std(collision_file['TS'][:, :, i], axis=0)
    ax.plot(mean_component, label="Média")
    ax.fill_between(np.arange(mean_component.size), mean_component - std_component, mean_component + std_component, alpha=0.5, label="Desvio Padrão")
    ax.legend()
    ax.set_xlabel("Snapshot")
    ax.set_ylabel(cn[i])
    fig.savefig(f"colisao_{cn2[i]}.pdf", dpi=300)
    plt.show()
```
